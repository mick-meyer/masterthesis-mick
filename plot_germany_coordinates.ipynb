```python
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np

# Load Natural Earth country boundaries (1:110m scale) directly from the URL
url = "https://naciscdn.org/naturalearth/110m/cultural/ne_110m_admin_0_countries.zip"
world = gpd.read_file(url)

# Select Germany from the global dataset
germany = world[world["ADMIN"] == "Germany"]

# Create a high-resolution figure (larger size and 300 DPI for crisp output)
fig, ax = plt.subplots(figsize=(10, 10), dpi=300)
world.plot(ax=ax, color="whitesmoke", edgecolor="gray")     # draw all countries
germany.plot(ax=ax, color="yellow", edgecolor="black")      # highlight Germany

# Set the map extent (lon/lat bounds) to focus on Germany and surroundings
ax.set_xlim(4, 17)
ax.set_ylim(46.5, 55.5)

# Add dashed guide lines (black) to frame the area of interest
ax.axhline(y=47.0, color="black", linestyle="--", linewidth=1)
ax.axhline(y=55.0, color="black", linestyle="--", linewidth=1)
ax.axvline(x=6.0,  color="black", linestyle="--", linewidth=1)
ax.axvline(x=15.0, color="black", linestyle="--", linewidth=1)

# Axis labels
ax.set_xlabel("Longitude (°E)")
ax.set_ylabel("Latitude (°N)")

# Tick marks: every 1° in longitude, every 0.5° in latitude
ax.set_xticks(np.arange(4, 18, 1))
ax.set_yticks(np.arange(46.5, 56, 0.5))

# Show grid and render the plot
plt.grid(True)
plt.show()

# Optional: save a high-resolution PNG of the figure
# fig.savefig("germany_map_highres.png", dpi=300, bbox_inches="tight")
```
