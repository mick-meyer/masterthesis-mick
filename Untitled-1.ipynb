```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
from sklearn.metrics import mean_squared_error

# Funktion zur Erzeugung von AR(1)-Noise
def generate_ar1_noise(length, std_vector, rho=0.8):
    noise = np.zeros(length)
    noise[0] = np.random.normal(0, std_vector[0])
    for t in range(1, length):
        epsilon = np.random.normal(0, std_vector[t])
        noise[t] = rho * noise[t - 1] + epsilon
    return noise

# Simulation für mehrere Prognosehorizonte
error_std = [[], [], [], [], []]
horizons_to_test = [48, 72, 96, 168, 336]

fig, axs = plt.subplots(2, 3, figsize=(18, 8))
axs = axs.flatten()

for k, i in enumerate(horizons_to_test):  
    real_wind = np.random.uniform(low=50, high=150, size=i)  # echte Windwerte

    horizons = np.array([0, 24, 48, 72, 96, 168, 336])
    mse_values = np.array([0, 0.499, 0.794, 0.899, 0.943, 1.011, 1.064])
    rmse_std = np.sqrt(mse_values)

    scaling_factor = np.std(real_wind)
    rmse_MW = rmse_std * scaling_factor

    interp_rmse = interp1d(horizons, rmse_MW, kind='cubic', fill_value='extrapolate')
    rmse_per_hour = interp_rmse(np.arange(0, i, 1))

    test = np.linspace(1.4, 1, 168)
    test = np.concatenate((test, np.ones(336 - 168)))

    noise = generate_ar1_noise(i, std_vector=test[:i] * rmse_per_hour, rho=0.8)
    synthetic_forecast = (real_wind + noise).clip(min=0)

    y_true_std = (real_wind - np.mean(real_wind)) / np.std(real_wind)
    y_pred_std = (synthetic_forecast - np.mean(real_wind)) / np.std(real_wind)
    mse_standardized = np.mean((y_true_std - y_pred_std) ** 2)
    error_std[k].append(mse_standardized)

    # Plot in Subplot
    ax = axs[k]
    ax.plot(real_wind, label='Echte Wind-Leistung', linewidth=2)
    ax.plot(synthetic_forecast, label='Prognose (AR1)', linestyle='--')
    ax.fill_between(range(i),
                    real_wind - rmse_per_hour,
                    real_wind + rmse_per_hour,
                    color='gray', alpha=0.2, label='RMSE-Band')
    ax.set_title(f'Horizont: {i}h')
    ax.set_xlabel('Stunde')
    ax.set_ylabel('Leistung (MW)')
    ax.grid(True)
    ax.legend()

# Leeres Feld ausblenden, falls weniger als 6 Plots
if len(horizons_to_test) < len(axs):
    for j in range(len(horizons_to_test), len(axs)):
        fig.delaxes(axs[j])

fig.suptitle('Synthetische Prognosepfade mit zeitlich korrelierten Fehlern (AR1)', fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# AR(1)-Fehlermodell
def generate_ar1_noise(length, std_vector, rho=0.8):
    noise = np.zeros(length)
    noise[0] = np.random.normal(0, std_vector[0])
    for t in range(1, length):
        epsilon = np.random.normal(0, std_vector[t])
        noise[t] = rho * noise[t - 1] + epsilon
    return noise

# Eingabe: real_cf muss extern definiert sein (z. B. 144h CF-Zeitreihe)
i = 144  # Horizont in Stunden

real_cf = n.generators_t.p_max_pu['DE0 0 offwind-dc-2035'][:144].values


# RMSE-Werte (z. B. aus LSTM-Modell)
horizons = np.array([0, 24, 48, 72, 96, 168, 336])
mse_values = np.array([0, 0.499, 0.794, 0.899, 0.943, 1.011, 1.064])
rmse_std = np.sqrt(mse_values)

# RMSE auf real_cf skalieren und interpolieren
scaling_factor = np.std(real_cf)
rmse_scaled = rmse_std * scaling_factor
interp_rmse = interp1d(horizons, rmse_scaled, kind='cubic', fill_value='extrapolate')
rmse_per_hour = interp_rmse(np.arange(i))

# Skalierungsvektor (leicht abfallend)
scale = np.linspace(1.4, 1, i)
std_vector = scale * rmse_per_hour
plt.figure(figsize=(12, 4))
for k in ['orange', 'green', 'black']:
# Fehler und Prognose erzeugen
    noise = generate_ar1_noise(i, std_vector=std_vector, rho=0.8)
    synthetic_forecast = np.clip(real_cf[:i] + noise, 0, 1)

    # Plot
    plt.plot(real_cf[:i], label='Real CF', linewidth=2)
    plt.plot(synthetic_forecast, label='Forecast CF (AR1)', linestyle='--', color=k)
    #plt.tight_layout()
    #plt.show()

plt.fill_between(range(i),
np.clip(real_cf[:i] - rmse_per_hour, 0, 1),
np.clip(real_cf[:i] + rmse_per_hour, 0, 1),
color='gray', alpha=0.2, label='RMSE-Band')
plt.xlabel('Stunde')
plt.ylabel('Capacity Factor')
plt.title('Synthetischer Forecast (144h) mit korreliertem Fehler')
plt.grid(True)
plt.legend()
```

```python
import numpy as np

# Beispiel: Deine echten cf-Werte (z. B. Länge n)
np.random.seed(42)
cf = np.random.rand(100)  # echte Werte zwischen 0 und 1

# Vorgegebene MSE-Werte für verschiedene Vorhersagehorizonte
horizons = np.array([0, 24, 48, 72, 96, 168, 336])
mse_values = np.array([0, 0.499, 0.794, 0.899, 0.943, 1.011, 1.064])

# Erzeuge Vorhersagen, die zu den MSE-Werten passen
predictions = []

for mse in mse_values:
    if mse == 0:
        # Perfekte Vorhersage
        pred = cf.copy()
    else:
        # Berechne Standardabweichung des Fehlers
        noise_std = np.sqrt(mse)
        # Füge Rauschen hinzu, sodass der MSE etwa stimmt
        noise = np.random.normal(loc=0, scale=noise_std, size=cf.shape)
        pred = cf + noise
        pred = np.clip(pred, 0, 1)  # Falls cf auf [0, 1] begrenzt ist
    predictions.append(pred)

# predictions ist ein Array der Form (len(horizons), len(cf))
predictions = np.array(predictions)

# Optional: überprüfe tatsächliche MSEs
actual_mse = np.mean((predictions - cf[None, :])**2, axis=1)
print("Gewünschte MSEs: ", mse_values)
print("Tatsächliche MSEs:", np.round(actual_mse, 3))
```
