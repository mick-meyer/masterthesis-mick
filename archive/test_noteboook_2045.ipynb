Graphen:
- wie verändern sich der marginal price am Strombus?
- wie verändert sich der marginal price am Wasserstoffbus?
- wie sieht die SOC des Wasserstoffspeichers aus?

```python
# SOC indem die Füllstände des H2 Stores aufsummiert werden und durch die installierte Kapazität geteilt werden

def get_hydrogen_soc(n):
    hydrogen_e = 0
    hydrogen_e_nom = 0

    for year in [2045, 2040, 2035, 2030, 2020]:

        hydrogen_e += n.stores_t.e[f"DE0 0 H2 Store-{year}"]
        hydrogen_e_nom += n.stores.e_nom[f"DE0 0 H2 Store-{year}"]

    soc = hydrogen_e / hydrogen_e_nom

    return soc
```

```python
fig, ax = plt.subplots(1,3, figsize=(16,9))

get_hydrogen_soc(n).plot(
    title="Hydrogen SOC Perfect Foresight",
    ax = ax[0]
)

get_hydrogen_soc(n_test).plot(
    title="Hydrogen SOC Rolling Horizon",
    ax = ax[1]
)

get_hydrogen_soc(n_test_rolling).plot(
    title="Hydrogen SOC Rolling Horizon 1000",
    ax = ax[2]
)
```
