```python
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np

# Natural Earth Ländergrenzen laden
url = "https://naciscdn.org/naturalearth/110m/cultural/ne_110m_admin_0_countries.zip"
world = gpd.read_file(url)

# Deutschland auswählen
germany = world[world["ADMIN"] == "Germany"]

# Plot mit hoher Auflösung
fig, ax = plt.subplots(figsize=(10,10), dpi=300)  # größer und 300 DPI
world.plot(ax=ax, color="whitesmoke", edgecolor="gray")
germany.plot(ax=ax, color="yellow", edgecolor="black")

# Grenzen setzen
ax.set_xlim(4, 17)
ax.set_ylim(46.5, 55.5)

# Linien einfügen (schwarz)
ax.axhline(y=47.0, color="black", linestyle="--", linewidth=1)
ax.axhline(y=55.0, color="black", linestyle="--", linewidth=1)
ax.axvline(x=6.0, color="black", linestyle="--", linewidth=1)
ax.axvline(x=15.0, color="black", linestyle="--", linewidth=1)

# Achsenbeschriftungen
ax.set_xlabel("Longitude (°E)")
ax.set_ylabel("Latitude (°N)")

# Ticks alle 1°
ax.set_xticks(np.arange(4, 18, 1))
ax.set_yticks(np.arange(46.5, 56, 0.5))

# Titel

plt.show()

# Optional speichern in hoher Auflösung:
# fig.savefig("germany_map_highres.png", dpi=300, bbox_inches='tight')
```
